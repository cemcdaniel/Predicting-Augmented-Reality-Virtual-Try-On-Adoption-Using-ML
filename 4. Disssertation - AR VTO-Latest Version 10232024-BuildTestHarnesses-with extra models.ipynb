## 1. Build initial Test Harness using original TAM Constructs as features w/out dropping 'outliers'

```python
### 1. Build initial Test Harness using original TAM Constructs as features w/out dropping 'outliers'
print('Build Harness Test including squared and cubed terms including outliers')
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score, KFold
from sklearn.linear_model import LinearRegression, Lasso, Ridge, ElasticNet
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error, explained_variance_score, max_error
from sklearn.preprocessing import StandardScaler
from sklearn.inspection import permutation_importance
import matplotlib.pyplot as plt
from xgboost import XGBRegressor

y = df_survey_cleaned['BI']
X = df_survey_cleaned[[
    'PU_centered', 'PEOU_centered', 'PE_centered', 'HM_centered', 'SOI_centered', 'SEI_centered', 'PPV_centered', 'AGE_num', 'Gender_num'
]]

# Get the original feature names
feature_names = X.columns.tolist()

# Create dummy variables for categorical variables
df_encoded = pd.get_dummies(X, columns=['AGE_num', 'Gender_num'], drop_first=True)

# Get the final feature names after encoding
final_feature_names = df_encoded.columns.tolist()

 
# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_encoded)



# Prepare models
models = [
    ('LR', LinearRegression()),
    ('Lasso', Lasso(alpha=.01, random_state=42)),
    ('Ridge', Ridge(alpha=1.0, random_state=42)),
    ('ElasticNet', ElasticNet(alpha=.01, l1_ratio=0.5, random_state=42)),
    ('KNN', KNeighborsRegressor()),
    ('RF', RandomForestRegressor(random_state=42)),
    ('SVR', SVR()),
    ('MLP', MLPRegressor(hidden_layer_sizes=(10), activation='relu', solver='adam',
                         alpha=0.0001, batch_size='auto', learning_rate='constant',
                         learning_rate_init=0.001, max_iter=1000, random_state=42)),
    ('XGBoost', XGBRegressor(random_state=42)),
    ('GradientBoosting', GradientBoostingRegressor(random_state=42))
]

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Prepare lists to store results
model_names = []
train_r2_list, test_r2_list = [], []
train_mse_list, test_mse_list = [], []
train_rmse_list, test_rmse_list = [], []
train_mae_list, test_mae_list = [], []
train_evs_list, test_evs_list = [], []
train_max_error_list, test_max_error_list = [], []
cv_r2_mean_list, cv_r2_std_list = [], []
feature_importance_list = []

# Get the original feature names
feature_names = X.columns.tolist()

# Evaluate each model
results = []
for name, model in models:
    # Train the model
    model.fit(X_train, y_train)
    
    # Make predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    
    # Calculate metrics
    train_r2 = r2_score(y_train, y_train_pred)
    test_r2 = r2_score(y_test, y_test_pred)
    train_mse = mean_squared_error(y_train, y_train_pred)
    test_mse = mean_squared_error(y_test, y_test_pred)
    train_rmse = np.sqrt(train_mse)
    test_rmse = np.sqrt(test_mse)
    train_mae = mean_absolute_error(y_train, y_train_pred)
    test_mae = mean_absolute_error(y_test, y_test_pred)
    train_evs = explained_variance_score(y_train, y_train_pred)
    test_evs = explained_variance_score(y_test, y_test_pred)
    train_max_error = max_error(y_train, y_train_pred)
    test_max_error = max_error(y_test, y_test_pred)
    
    # Perform cross-validation
    kfold = KFold(n_splits=10, random_state=42, shuffle=True)
    cv_results = cross_val_score(model, X_scaled, y, cv=kfold, scoring='r2')

    # Calculate feature importance
    if hasattr(model, 'feature_importances_'):
        importances = model.feature_importances_
    elif hasattr(model, 'coef_'):
        importances = np.abs(model.coef_)
        if isinstance(importances, np.ndarray) and importances.ndim > 1:
            importances = importances.flatten()
    else:
        # For models without built-in feature importance, use permutation importance
        perm_importance = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=42)
        importances = perm_importance.importances_mean
    
    # Get top feature importances
    top_features = pd.DataFrame({'feature': final_feature_names, 'importance': importances})
    top_features = top_features.sort_values('importance', ascending=False)
    
    # Handle cases where all importances are zero (like in Lasso with high regularization)
    if top_features['importance'].sum() == 0:
        print("  All feature importances are zero.")
    else:
        top_features = top_features[top_features['importance'] > 0].head(20)
        
        if len(top_features) == 0:
            print("  No non-zero feature importances found.")
        else:
            print("  Top 5 Features:")
            print(top_features)

    top_features = top_features.sort_values('importance', ascending=False).head(15)
    
    # Store results
    results.append(cv_results)
    model_names.append(name)
    train_r2_list.append(train_r2)
    test_r2_list.append(test_r2)
    train_mse_list.append(train_mse)
    test_mse_list.append(test_mse)
    train_rmse_list.append(train_rmse)
    test_rmse_list.append(test_rmse)
    train_mae_list.append(train_mae)
    test_mae_list.append(test_mae)
    train_evs_list.append(train_evs)
    test_evs_list.append(test_evs)
    train_max_error_list.append(train_max_error)
    test_max_error_list.append(test_max_error)
    cv_r2_mean_list.append(cv_results.mean())
    cv_r2_std_list.append(cv_results.std())
    feature_importance_list.append(top_features)
    
    # Print results
    print(f"{name}:")
    print(f"  Training R^2: {train_r2:.4f}")
    print(f"  Testing R^2: {test_r2:.4f}")
    print(f"  Training MSE: {train_mse:.4f}")
    print(f"  Testing MSE: {test_mse:.4f}")
    print(f"  Training RMSE: {train_rmse:.4f}")
    print(f"  Testing RMSE: {test_rmse:.4f}")
    print(f"  Training MAE: {train_mae:.4f}")
    print(f"  Testing MAE: {test_mae:.4f}")
    print(f"  Training EVS: {train_evs:.4f}")
    print(f"  Testing EVS: {test_evs:.4f}")
    print(f"  Training Max Error: {train_max_error:.4f}")
    print(f"  Testing Max Error: {test_max_error:.4f}")
    print(f"  Cross-validation R^2: {cv_results.mean():.4f} (+/- {cv_results.std() * 2:.4f})")
    print("  Top 5 Features:")
    print(top_features)
    print()


# Create DataFrame with results
df_results_1 = pd.DataFrame({
    'Model': model_names,
    'Training R^2': train_r2_list,
    'Testing R^2': test_r2_list,
    'Training MSE': train_mse_list,
    'Testing MSE': test_mse_list,
    'Training RMSE': train_rmse_list,
    'Testing RMSE': test_rmse_list,
    'Training MAE': train_mae_list,
    'Testing MAE': test_mae_list,
    'Training EVS': train_evs_list,
    'Testing EVS': test_evs_list,
    'Training Max Error': train_max_error_list,
    'Testing Max Error': test_max_error_list,
    'CV R^2 Mean': cv_r2_mean_list,
    'CV R^2 Std': cv_r2_std_list,
})

# Calculate the CV R^2 range
df_results_1['CV R^2 Range'] = df_results_1.apply(lambda row: f"{row['CV R^2 Mean']:.4f} (+/- {row['CV R^2 Std']*2:.4f})", axis=1)

# List of columns to rename
columns_to_rename = ['R^2', 'MSE', 'RMSE', 'MAE', 'EVS', 'Max Error', 'CV R^2 Range']

# Create a dictionary for renaming
rename_dict = {f'Training {col}': f'P_Training_{col}' for col in columns_to_rename}
rename_dict.update({f'Testing {col}': f'P_Testing_{col}' for col in columns_to_rename})
rename_dict['CV R^2 Range'] = 'P_CV R^2 Range'

# Rename the columns
df_results_1 = df_results_1.rename(columns=rename_dict)

# Add feature importance separately
df_results_1['Top Features'] = feature_importance_list

# Display the DataFrame
#print(df_results_1)

# Create a DataFrame to hold feature importances for all models
feature_importance_df = pd.DataFrame(index=final_feature_names)

for name, top_features in zip(model_names, feature_importance_list):
    feature_importance_df[name] = 0
    for _, row in top_features.iterrows():
        feature_importance_df.loc[row['feature'], name] = row['importance']


# Select top N features across all models
N = 20  # You can adjust this number
top_features = feature_importance_df.sum(axis=1).nlargest(N).index

# Create a new DataFrame with only the top features
top_features_df = feature_importance_df.loc[top_features]

# Create grouped bar plot
plt.figure(figsize=(15, 10))
bar_width = 0.1
index = np.arange(len(top_features))

for i, model in enumerate(model_names):
    plt.bar(index + i*bar_width, 
            feature_importance_df.loc[top_features, model], 
            bar_width, 
            label=model)

plt.xlabel('Features')
plt.ylabel('Importance')
plt.title('Top Feature Importances Across Models')
plt.xticks(index + bar_width * (len(model_names) - 1) / 2, top_features, rotation=45, ha='right')
plt.legend(title='Models', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()


# Create a binary DataFrame indicating which models use each feature
model_usage_df = (top_features_df != 0).astype(int)

# Combine importance values and usage indicators
combined_df = pd.DataFrame()
for model in model_names:
    combined_df[f"{model}_Importance"] = top_features_df[model]
    combined_df[f"{model}_Usage"] = model_usage_df[model]

# Sort the DataFrame by the sum of importances across all models
combined_df = combined_df.sort_values(by=[f"{model}_Importance" for model in model_names], ascending=False)

# Display the table
#print(combined_df)


# Create boxplot
plt.figure(figsize=(10, 4))
plt.title('Algorithm Comparison all features of R^2 (10-Fold CV) Using Interaction and Polynomial Features Including Outliers')
plt.boxplot(results)
plt.xticks(range(1, len(model_names) + 1), model_names, rotation=45, ha='right')
plt.ylabel('R^2 Score')
plt.tight_layout()
plt.show()

# Function to create spider plot
def create_spider_plot(data, title, metrics):
    # Prepare the data for plotting
    values = data[[f'P_{m}' for m in metrics]].values

    # Normalize the metrics
    def normalize(x):
        return (x - np.min(x)) / (np.max(x) - np.min(x))

    values_norm = np.apply_along_axis(normalize, 0, values)

    # Set up the angles for each metric
    angles = np.linspace(0, 2*np.pi, len(metrics), endpoint=False)

    # Close the plot by appending the first value to the end
    values_norm = np.c_[values_norm, values_norm[:, 0]]
    angles = np.concatenate((angles, [angles[0]]))

    # Create the plot
    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(projection='polar'))

    # Plot each model
    for i, model in enumerate(model_names):
        ax.plot(angles, values_norm[i], 'o-', linewidth=2, label=model)
        ax.fill(angles, values_norm[i], alpha=0.1)

    # Set the labels
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(metrics, fontsize=10)

    # Add legend
    plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))

    # Add title
    plt.title(title, size=16, y=1.1)

    # Show the plot
    plt.tight_layout()
    plt.show()

# Create spider plots
train_metrics = ['Training_R^2', 'Training_MSE', 'Training_RMSE', 'Training_MAE', 'Training_EVS', 'Training_Max Error']
test_metrics = ['Testing_R^2', 'Testing_MSE', 'Testing_RMSE', 'Testing_MAE', 'Testing_EVS', 'Testing_Max Error']

create_spider_plot(df_results_1, "All Features (Interaction and Polynomial) Model Performance Comparison (Train Metrics)", train_metrics)
create_spider_plot(df_results_1, "All Features (Interaction and Polynomial) Model Performance Comparison (Test Metrics)", test_metrics)

# Plot feature importance for each model
for name, top_features in zip(model_names, feature_importance_list):
    plt.figure(figsize=(8, 4))
    plt.bar(top_features['feature'], top_features['importance'])
    plt.title(f'All Features (Interaction and Polynomial) -Top Feature Importances for {name}')
    plt.xlabel('Features')
    plt.ylabel('Importance')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

print('End of Script 1')
```

```python
### print ('2. Build initial Test Harness using PCA Feature list')
 
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score, KFold
from sklearn.linear_model import LinearRegression, Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error, explained_variance_score, max_error
from sklearn.preprocessing import StandardScaler
from sklearn.inspection import permutation_importance
import matplotlib.pyplot as plt
from xgboost import XGBRegressor

import json

# Open and read the JSON file
try:
    with open('PCA_selected_features.json', 'r') as json_file:
        feature_data = json.load(json_file)

    # Extract the list of selected features
    pca_selected_features = feature_data['PCA_selected_features']

    print("Type of pca_selected_features:", type(pca_selected_features))
    print("First few items in pca_selected_features:")
    for item in pca_selected_features[:5]:
        print(f"  {item} (type: {type(item)})")

    # If pca_selected_features is a list of strings, use this:
    selected_features = list(dict.fromkeys(pca_selected_features))

    print(f"\nSuccessfully loaded {len(feature_vars)} unique PCA selected features.")
    print("\nFirst 5 features:")
    for feature in selected_features[:5]:
        print(f"- {feature}")

    print(f"\nLast 5 features:")
    for feature in selected_features[-5:]:
        print(f"- {feature}")

    print(f"\nNumber of unique features: {len(feature_vars)}")

except FileNotFoundError:
    print("Error: The file 'PCA_selected_features.json' was not found.")
except json.JSONDecodeError:
    print("Error: There was a problem decoding the JSON file.")
except KeyError:
    print("Error: The key 'PCA_selected_features' was not found in the JSON file.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
    
# Select the first 30 features from feature_vars
feature_vars = selected_features[:30]

print(f"Number of features selected: {len(feature_vars)}")
print("Selected features:")
print(feature_vars)


# Use these selected features to create X
X = df_survey_cleaned[feature_vars]

# Get the original feature names
feature_names = X.columns.tolist()

# Create dummy variables for categorical variables
#df_encoded = pd.get_dummies(X, columns=['AGE_num', 'Gender_num'], drop_first=True)

# Get the final feature names after encoding
#final_feature_names = df_encoded.columns.tolist()

 
# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)



# Prepare models
models = [
    ('LR', LinearRegression()),
    ('Lasso', Lasso(alpha=.01, random_state=42)),
    ('Ridge', Ridge(alpha=.01, random_state=42)),
    ('ElasticNet', ElasticNet(alpha=.01, l1_ratio=0.5, random_state=42)),
    ('KNN', KNeighborsRegressor()),
    ('RF', RandomForestRegressor(random_state=42)),
    ('SVR', SVR()),
    ('MLP', MLPRegressor(hidden_layer_sizes=(10), activation='relu', solver='adam',
                         alpha=0.0001, batch_size='auto', learning_rate='constant',
                         learning_rate_init=0.001, max_iter=1000, random_state=42)),
    ('XGBoost', XGBRegressor(random_state=42)),
    ('GradientBoosting', GradientBoostingRegressor(random_state=42))
]

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Prepare lists to store results
model_names = []
train_r2_list, test_r2_list = [], []
train_mse_list, test_mse_list = [], []
train_rmse_list, test_rmse_list = [], []
train_mae_list, test_mae_list = [], []
train_evs_list, test_evs_list = [], []
train_max_error_list, test_max_error_list = [], []
cv_r2_mean_list, cv_r2_std_list = [], []
feature_importance_list = []

# Get the original feature names
feature_names = X.columns.tolist()

print(f"ffeature_names: {feature_names}")

# Evaluate each model
results = []
for name, model in models:
    # Train the model
    model.fit(X_train, y_train)
    
    # Make predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    
    # Calculate metrics
    train_r2 = r2_score(y_train, y_train_pred)
    test_r2 = r2_score(y_test, y_test_pred)
    train_mse = mean_squared_error(y_train, y_train_pred)
    test_mse = mean_squared_error(y_test, y_test_pred)
    train_rmse = np.sqrt(train_mse)
    test_rmse = np.sqrt(test_mse)
    train_mae = mean_absolute_error(y_train, y_train_pred)
    test_mae = mean_absolute_error(y_test, y_test_pred)
    train_evs = explained_variance_score(y_train, y_train_pred)
    test_evs = explained_variance_score(y_test, y_test_pred)
    train_max_error = max_error(y_train, y_train_pred)
    test_max_error = max_error(y_test, y_test_pred)
    
    # Perform cross-validation
    kfold = KFold(n_splits=10, random_state=42, shuffle=True)
    cv_results = cross_val_score(model, X_scaled, y, cv=kfold, scoring='r2')

    # Calculate feature importance
    if hasattr(model, 'feature_importances_'):
        importances = model.feature_importances_
    elif hasattr(model, 'coef_'):
        importances = np.abs(model.coef_)
        if isinstance(importances, np.ndarray) and importances.ndim > 1:
            importances = importances.flatten()
    else:
        # For models without built-in feature importance, use permutation importance
        perm_importance = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=42)
        importances = perm_importance.importances_mean
    
    # Get top feature importances
        top_features = pd.DataFrame({'feature': feature_names, 'importance': importances})
        top_features = top_features.sort_values('importance', ascending=False)
    
    # Handle cases where all importances are zero (like in Lasso with high regularization)
    if top_features['importance'].sum() == 0:
        print("  All feature importances are zero.")
    else:
        top_features = top_features[top_features['importance'] > 0].head(20)
        
        if len(top_features) == 0:
            print("  No non-zero feature importances found.")
        else:
            print("  Top 5 Features:")
            print(top_features)

    top_features = top_features.sort_values('importance', ascending=False).head(15)
    
    # Store results
    results.append(cv_results)
    model_names.append(name)
    train_r2_list.append(train_r2)
    test_r2_list.append(test_r2)
    train_mse_list.append(train_mse)
    test_mse_list.append(test_mse)
    train_rmse_list.append(train_rmse)
    test_rmse_list.append(test_rmse)
    train_mae_list.append(train_mae)
    test_mae_list.append(test_mae)
    train_evs_list.append(train_evs)
    test_evs_list.append(test_evs)
    train_max_error_list.append(train_max_error)
    test_max_error_list.append(test_max_error)
    cv_r2_mean_list.append(cv_results.mean())
    cv_r2_std_list.append(cv_results.std())
    feature_importance_list.append(top_features)
    
    # Print results
    print(f"{name}:")
    print(f"  Training R^2: {train_r2:.4f}")
    print(f"  Testing R^2: {test_r2:.4f}")
    print(f"  Training MSE: {train_mse:.4f}")
    print(f"  Testing MSE: {test_mse:.4f}")
    print(f"  Training RMSE: {train_rmse:.4f}")
    print(f"  Testing RMSE: {test_rmse:.4f}")
    print(f"  Training MAE: {train_mae:.4f}")
    print(f"  Testing MAE: {test_mae:.4f}")
    print(f"  Training EVS: {train_evs:.4f}")
    print(f"  Testing EVS: {test_evs:.4f}")
    print(f"  Training Max Error: {train_max_error:.4f}")
    print(f"  Testing Max Error: {test_max_error:.4f}")
    print(f"  Cross-validation R^2: {cv_results.mean():.4f} (+/- {cv_results.std() * 2:.4f})")
    print("  Top 5 Features:")
    print(top_features)
    print()


# Create DataFrame with results
df_results_2 = pd.DataFrame({
    'Model': model_names,
    'Training R^2': train_r2_list,
    'Testing R^2': test_r2_list,
    'Training MSE': train_mse_list,
    'Testing MSE': test_mse_list,
    'Training RMSE': train_rmse_list,
    'Testing RMSE': test_rmse_list,
    'Training MAE': train_mae_list,
    'Testing MAE': test_mae_list,
    'Training EVS': train_evs_list,
    'Testing EVS': test_evs_list,
    'Training Max Error': train_max_error_list,
    'Testing Max Error': test_max_error_list,
    'CV R^2 Mean': cv_r2_mean_list,
    'CV R^2 Std': cv_r2_std_list,
})

# Calculate the CV R^2 range
df_results_2['CV R^2 Range'] = df_results_2.apply(lambda row: f"{row['CV R^2 Mean']:.4f} (+/- {row['CV R^2 Std']*2:.4f})", axis=1)

# List of columns to rename
columns_to_rename = ['R^2', 'MSE', 'RMSE', 'MAE', 'EVS', 'Max Error', 'CV R^2 Range']

# Create a dictionary for renaming
rename_dict = {f'Training {col}': f'P_Training_{col}' for col in columns_to_rename}
rename_dict.update({f'Testing {col}': f'P_Testing_{col}' for col in columns_to_rename})
rename_dict['CV R^2 Range'] = 'P_CV R^2 Range'

# Rename the columns
df_results_2 = df_results_2.rename(columns=rename_dict)

# Add feature importance separately
df_results_2['Top Features'] = feature_importance_list

# Display the DataFrame
print(df_results_2)

# Check and align lengths
min_length = min(len(feature_names), len(importances))
aligned_features = feature_names[:min_length]
aligned_importances = importances[:min_length]

# Create a DataFrame to hold feature importances for all models
feature_importance_df = pd.DataFrame(index=feature_names)

for name, top_features in zip(model_names, feature_importance_list):
    feature_importance_df[name] = 0
    for _, row in top_features.iterrows():
        feature_importance_df.loc[row['feature'], name] = row['importance']


# Select top N features across all models
N = 20  
top_features = feature_importance_df.sum(axis=1).nlargest(N).index

# Create a new DataFrame with only the top features
top_features_df = feature_importance_df.loc[top_features]


# Create grouped bar plot
plt.figure(figsize=(15, 10))
bar_width = 0.1
index = np.arange(len(top_features))

for i, model in enumerate(model_names):
    plt.bar(index + i*bar_width, 
            feature_importance_df.loc[top_features, model], 
            bar_width, 
            label=model)

plt.xlabel('Features')
plt.ylabel('Importance')
plt.title('Top Feature Importances Across Models')
plt.xticks(index + bar_width * (len(model_names) - 1) / 2, top_features, rotation=45, ha='right')
plt.legend(title='Models', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()


# Create a binary DataFrame indicating which models use each feature
model_usage_df = (top_features_df != 0).astype(int)

# Combine importance values and usage indicators
combined_df = pd.DataFrame()
for model in model_names:
    combined_df[f"{model}_Importance"] = top_features_df[model]
    combined_df[f"{model}_Usage"] = model_usage_df[model]

# Sort the DataFrame by the sum of importances across all models
combined_df = combined_df.sort_values(by=[f"{model}_Importance" for model in model_names], ascending=False)

# Display the table
#print(combined_df)


# Create boxplot
plt.figure(figsize=(10, 4))
plt.title('Algorithm Comparison all features of R^2 (10-Fold CV) Using PCA Features')
plt.boxplot(results)
plt.xticks(range(1, len(model_names) + 1), model_names, rotation=45, ha='right')
plt.ylabel('R^2 Score')
plt.tight_layout()
plt.show()

# Function to create spider plot
def create_spider_plot(data, title, metrics):
    # Prepare the data for plotting
    values = data[[f'P_{m}' for m in metrics]].values

    # Normalize the metrics
    def normalize(x):
        return (x - np.min(x)) / (np.max(x) - np.min(x))

    values_norm = np.apply_along_axis(normalize, 0, values)

    # Set up the angles for each metric
    angles = np.linspace(0, 2*np.pi, len(metrics), endpoint=False)

    # Close the plot by appending the first value to the end
    values_norm = np.c_[values_norm, values_norm[:, 0]]
    angles = np.concatenate((angles, [angles[0]]))

    # Create the plot
    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(projection='polar'))

    # Plot each model
    for i, model in enumerate(model_names):
        ax.plot(angles, values_norm[i], 'o-', linewidth=2, label=model)
        ax.fill(angles, values_norm[i], alpha=0.1)

    # Set the labels
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(metrics, fontsize=10)

    # Add legend
    plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))

    # Add title
    plt.title(title, size=16, y=1.1)

    # Show the plot
    plt.tight_layout()
    plt.show()

# Create spider plots
train_metrics = ['Training_R^2', 'Training_MSE', 'Training_RMSE', 'Training_MAE', 'Training_EVS', 'Training_Max Error']
test_metrics = ['Testing_R^2', 'Testing_MSE', 'Testing_RMSE', 'Testing_MAE', 'Testing_EVS', 'Testing_Max Error']

create_spider_plot(df_results_1, "All Features (Interaction and Polynomial) Model Performance Comparison (Train Metrics)", train_metrics)
create_spider_plot(df_results_1, "All Features (Interaction and Polynomial) Model Performance Comparison (Test Metrics)", test_metrics)

# Plot feature importance for each model
for name, top_features in zip(model_names, feature_importance_list):
    plt.figure(figsize=(8, 4))
    plt.bar(top_features['feature'], top_features['importance'])
    plt.title(f'All Features (Interaction and Polynomial) -Top Feature Importances for {name}')
    plt.xlabel('Features')
    plt.ylabel('Importance')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

print('End of Script 2')

```
