```python
import pandas as pd
import numpy as np
from semopy import Model, calc_stats
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt

# Load your dataset
df_survey_cleaned = pd.read_csv(r'C:\Users\cemcd\Documents\AR VTO Dissertation\AR VTO Dissertation Oct 6\AR VTO Latest Version Oct16 using PCA as Input\df_survey_cleaned.csv')

# Define the TAM model specification
model_spec = """
# Measurement model
PU =~ PU1 + PU2 + PU3 + PU4
PEOU =~ PEOU1 + PEOU2 + PEOU3
BI =~ BI1 + BI2 + BI3
PPV =~ PPV1 + PPV2 + PPV3
SOI =~ SOI1 + SOI2 + SOI3 + SOI4
SEI =~ SEI1 + SEI2 + SEI3
HM =~ HM1 + HM2 + HM3


# Structural model
PEOU ~ SOI + SEI + AGE_num + Gender_num
PU ~ HM + PPV + SOI + SEI + AGE_num + Gender_num
HM ~ PEOU
BI ~ AGEPEOU_centered
BI ~ GenderPEOU_centered
BI ~ AGEPU_centered
BI ~ GenderPU_centered
BI ~ PU
BI ~ PEOU
BI ~ HM
"""

# Initialize KFold and lists to store estimates and metrics from each fold
k = 5
kf = KFold(n_splits=k, shuffle=True, random_state=42)
all_estimates = []
train_mse, train_r2, test_mse, test_r2 = [], [], [], []
cri_values, tli_values, bic_values, aic_values = [], [], [], []
# Initialize metric lists
train_mse = []
train_r2 = []
train_mae = []
train_evs = []
test_mse = []
test_r2 = []
test_mae = []
test_evs = []

# Function to calculate CRI
def calculate_cri(model, latent_variable):
    loadings = model.inspect(latent_variable, 'loading')
    if loadings is None or loadings.empty:
        print(f"Warning: No loadings found for {latent_variable}")
        return None
    squared_sum = np.sum(loadings['Estimate'])**2
    sum_squared_errors = np.sum(1 - loadings['Estimate']**2)
    return squared_sum / (squared_sum + sum_squared_errors)

# Perform k-fold cross-validation
for fold, (train_index, test_index) in enumerate(kf.split(df_survey_cleaned), 1):
    print(f"\nFold {fold}")
    
    # Split the data
    train_data = df_survey_cleaned.iloc[train_index]
    test_data = df_survey_cleaned.iloc[test_index]
    
    # Fit the SEM model on training data
    model = Model(model_spec)
    res = model.fit(train_data)
    
    # Extract path coefficients and store them
    estimates = model.inspect()
    all_estimates.append(estimates)
    
    # Calculate additional metrics
    stats = calc_stats(model)
    cri_list = [calculate_cri(model, lv) for lv in ['PU', 'PEOU', 'BI', 'PPV', 'SOI', 'SEI', 'HM']]
    cri_list = [cri for cri in cri_list if cri is not None]  # Remove None values
    if cri_list:
        cri_values.append(np.mean(cri_list))
    else:
        print("Warning: No valid CRI values calculated for this fold")
    tli_values.append(stats['TLI'])
    bic_values.append(stats['BIC'])
    aic_values.append(stats['AIC'])
    
    # Calculate BI score for train and test data
    coef_AGEPEOU_BI = estimates.loc[(estimates['lval'] == 'BI') & (estimates['rval'] == 'AGEPEOU_centered'), 'Estimate'].values[0]
    coef_GenderPEOU_BI = estimates.loc[(estimates['lval'] == 'BI') & (estimates['rval'] == 'GenderPEOU_centered'), 'Estimate'].values[0]
    coef_AGEPU_BI = estimates.loc[(estimates['lval'] == 'BI') & (estimates['rval'] == 'AGEPU_centered'), 'Estimate'].values[0]
    coef_GenderPU_BI = estimates.loc[(estimates['lval'] == 'BI') & (estimates['rval'] == 'GenderPU_centered'), 'Estimate'].values[0]
    coef_PU_BI = estimates.loc[(estimates['lval'] == 'BI') & (estimates['rval'] == 'PU'), 'Estimate'].values[0]
    coef_PEOU_BI = estimates.loc[(estimates['lval'] == 'BI') & (estimates['rval'] == 'PEOU'), 'Estimate'].values[0]
    coef_HM_BI = estimates.loc[(estimates['lval'] == 'BI') & (estimates['rval'] == 'HM'), 'Estimate'].values[0]

    for data in [train_data, test_data]:
        data.loc[:, 'BI_score'] = (
            coef_AGEPEOU_BI * data['AGEPEOU_centered'] +
            coef_GenderPEOU_BI * data['GenderPEOU_centered'] +
            coef_AGEPU_BI * data['AGEPU_centered'] +
            coef_GenderPU_BI * data['GenderPU_centered'] +
            coef_PU_BI * data['PU'] +
            coef_PEOU_BI * data['PEOU'] +
            coef_HM_BI * data['HM']
        )
    
    # Calculate MSE and R-squared for this fold
    from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, explained_variance_score
    # Calculate metrics for this fold
    train_mse.append(mean_squared_error(train_data['BI'], train_data['BI_score']))
    train_r2.append(r2_score(train_data['BI'], train_data['BI_score']))
    train_mae.append(mean_absolute_error(train_data['BI'], train_data['BI_score']))
    train_evs.append(explained_variance_score(train_data['BI'], train_data['BI_score']))

    test_mse.append(mean_squared_error(test_data['BI'], test_data['BI_score']))
    test_r2.append(r2_score(test_data['BI'], test_data['BI_score']))
    test_mae.append(mean_absolute_error(test_data['BI'], test_data['BI_score']))
    test_evs.append(explained_variance_score(test_data['BI'], test_data['BI_score']))
    

# Calculate average path coefficients across all folds
avg_coefficients = {
    'AGEPEOU_centered': np.mean([est.loc[(est['lval'] == 'BI') & (est['rval'] == 'AGEPEOU_centered'), 'Estimate'].values[0] for est in all_estimates]),
    'GenderPEOU_centered': np.mean([est.loc[(est['lval'] == 'BI') & (est['rval'] == 'GenderPEOU_centered'), 'Estimate'].values[0] for est in all_estimates]),
    'AGEPU_centered': np.mean([est.loc[(est['lval'] == 'BI') & (est['rval'] == 'AGEPU_centered'), 'Estimate'].values[0] for est in all_estimates]),
    'GenderPU_centered': np.mean([est.loc[(est['lval'] == 'BI') & (est['rval'] == 'GenderPU_centered'), 'Estimate'].values[0] for est in all_estimates]),
    'PU': np.mean([est.loc[(est['lval'] == 'BI') & (est['rval'] == 'PU'), 'Estimate'].values[0] for est in all_estimates]),
    'PEOU': np.mean([est.loc[(est['lval'] == 'BI') & (est['rval'] == 'PEOU'), 'Estimate'].values[0] for est in all_estimates]),
    'HM': np.mean([est.loc[(est['lval'] == 'BI') & (est['rval'] == 'HM'), 'Estimate'].values[0] for est in all_estimates])
}

# Calculate average metrics
avg_train_mse = np.mean(train_mse)
avg_train_r2 = np.mean(train_r2)
avg_train_mae = np.mean(train_mae)
avg_train_evs = np.mean(train_evs)
avg_test_mse = np.mean(test_mse)
avg_test_r2 = np.mean(test_r2)
avg_test_mae = np.mean(test_mae)
avg_test_evs = np.mean(test_evs)

# Create a DataFrame for feature importance
features_importance = pd.DataFrame(avg_coefficients.items(), columns=['Feature', 'Coefficient'])
features_importance['Absolute Coefficient'] = features_importance['Coefficient']

# Sort by absolute coefficient values
features_importance_sorted = features_importance.sort_values(by='Absolute Coefficient', ascending=False)

print("\nTAM Feature Importance based on Average Path Coefficients:")
print(features_importance_sorted)

# Visualize feature importance
plt.figure(figsize=(12, 6))
plt.barh(features_importance_sorted['Feature'], features_importance_sorted['Absolute Coefficient'], color='skyblue')
plt.xlabel('Absolute Coefficient Value')
plt.title('Feature Importance based on TAM Model Coefficients')
plt.grid(axis='x')
plt.tight_layout()
plt.show()

# Convert lists to numpy arrays for easier calculations
train_mse, train_r2, test_mse, test_r2 = map(np.array, [train_mse, train_r2, test_mse, test_r2])
cri_values, tli_values, bic_values, aic_values = map(np.array, [cri_values, tli_values, bic_values, aic_values])


# Print Semopy TAM Regressor results
print("\nSemopy TAM Regressor (Best Model) Results (5-fold Cross-Validation):")
print("--------------------------------")
print(f"Training MSE: {avg_train_mse:.4f}")
print(f"Training R^2: {avg_train_r2:.4f} ")
print(f"Testing MSE: {avg_test_mse:.4f}")
print(f"Testing R^2: {avg_test_r2:.4f}")
print(f"Testing MAE: {avg_test_mae:.4f}")
print(f"Testing EVS: {avg_test_evs:.4f}")
print(f"CRI: {cri_values.mean():.4f}")
print(f"TLI: {tli_values.mean():.4f}")
print(f"BIC: {bic_values.mean():.4f}")
print(f"AIC: {aic_values.mean():.4f}")


# Create model_metrics dataframe
model_metrics = pd.DataFrame({
    'Model Name': ['Semopy TAM Model'],
    'Run Type': ['Initial Run'],
    'Train MSE': [avg_train_mse],
   # 'Train MSE Std': [train_mse_std],
    'Train R2': [avg_train_r2],
   # 'Train R2 Std': [train_r2_std],
    'Test MSE': [avg_test_mse],
   # 'Test MSE Std': [avg_test_mse_std],
    'Test R2': [avg_test_r2],
  #  'Test R2 Std': [avg_test_r2_std],
    'Train MAE': [avg_train_mae],
  #  'Train MAE Std': avg_[train_mae_std],
    'Test MAE': [avg_test_mae],
  #  'Test MAE Std': [avg_test_mae_std],
    'Train EVS': [avg_train_evs],
  #  'Train EVS Std': [avg_train_evs_std],
    'Test EVS': [avg_test_evs],
 #   'Test EVS Std': [test_evs_std],
    'cri_mean': [cri_values.mean()],
  #  'cri_std': [cri_values.std()],
    'tli_mean': [tli_values.mean()],
  #  'tli_std': [tli_values.std()],
    'bic_mean': [bic_values.mean()],
   # 'bic_std': [bic_values.std()],
    'aic_mean': [aic_values.mean()],
   # 'aic_std': [aic_values.std()]
})


print(f"Model Metrics: {model_metrics}")

model_metrics.to_csv('TAM_model_metrics.csv', index=False)

print('End of TAM script')
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, explained_variance_score
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore', module='statsmodels')

#read csv file
import pandas as pd
df_survey_cleaned = pd.read_csv('df_survey_cleaned.csv')

feature_vars = ['GenderSEI_centered','HM_centered','AGESEI_centered','GenderHM_centered','GenderPU_centered','GenderPPV_centered',
                'HMSEI_centered_squared','HMAT_centered','HMPE_centered']

# Assuming df_survey_cleaned is your dataframe
y = df_survey_cleaned['BI']
X = df_survey_cleaned[feature_vars]

# Add a constant term to the independent variables
X = sm.add_constant(X)

# Fit the OLS model
results = sm.OLS(y, X).fit()

# Print model summary results
print(results.summary())

# Get the predicted values
y_pred = results.predict(X)

# Calculate metrics
r_squared = results.rsquared
adj_r_squared = results.rsquared_adj
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y, y_pred)
evs = explained_variance_score(y, y_pred)

print(f"R Squared (R^2): {r_squared:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"Explained Variance Score (EVS): {evs:.4f}")

# Create a dictionary with the results
statsmodel_metrics = {
    'Model Name': 'StatsModel OLS',
    'Train R2': r_squared,
    'Test R2': r_squared,  # Same as Train R2 for this model
    'Train MSE': mse,
    'Test MSE': mse,  # Same as Train MSE for this model
    'Train RMSE': rmse,
    'Test RMSE': rmse,  # Same as Train RMSE for this model
    'Train MAE': mae,
    'Test MAE': mae,  # Same as Train MAE for this model
    'Train EVS': evs,
    'Test EVS': evs,  # Same as Train EVS for this model
    'CV R2 Mean': r_squared,  # Not applicable for this model, using R2
    'CV R2 Std': 0,  # Not applicable for this model
    'CV MSE Mean': mse,  # Not applicable for this model, using MSE
    'CV MSE Std': 0,  # Not applicable for this model
    'Adjusted R-squared': adj_r_squared,
    'Log-Likelihood': results.llf,
    'AIC': results.aic,
    'BIC': results.bic
}

# Add statsmodel metrics to all_model_metrics
all_model_metrics.append(statsmodel_metrics)

# Create a DataFrame with all model metrics including statsmodel
all_models_df = pd.DataFrame(all_model_metrics)

# Save the combined metrics to a CSV file
all_models_df.to_csv('all_models_metrics_with_statsmodel.csv', index=False)

print("Combined metrics for all models including statsmodel saved to 'all_models_metrics_with_statsmodel.csv'")

# Display the combined metrics
print(all_models_df)

# Diagnostic Plots
# 1. Q-Q Plot
fig = sm.qqplot(results.resid, line='45', fit=True)
plt.title('Q-Q Plot of Residuals')
plt.show()

# 2. Partial Regression Plots
fig = sm.graphics.plot_partregress_grid(results, fig=plt.figure(figsize=(20, 20)))
plt.show()

# 3. Influence Plot
fig, ax = plt.subplots(figsize=(12, 8))
sm.graphics.influence_plot(results, ax=ax, criterion="cooks")
plt.title("Influence Plot")
plt.show()

# 4. Leverage-Residual Plot
fig, ax = plt.subplots(figsize=(12, 8))
sm.graphics.plot_leverage_resid2(results, ax=ax)
plt.title("Leverage-Residual Plot")
plt.show()

# 5. Correlation Heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(df_survey_cleaned[feature_vars + ['BI']].corr(), 
            annot=True, 
            cmap='coolwarm', 
            vmin=-1, 
            vmax=1, 
            center=0,
            annot_kws={'size': 8})
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

print('End of STATS Model analysis')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.stats.stattools import durbin_watson
from statsmodels.graphics.gofplots import ProbPlot

def test_ols_assumptions(model, X, y):
    # 1. Linearity
    plt.figure(figsize=(10, 6))
    plt.scatter(model.predict(), model.resid)
    plt.xlabel("Predicted values")
    plt.ylabel("Residuals")
    plt.title("Residuals vs Predicted values")
    plt.show()
    
    # 2. Independence (Durbin-Watson test)
    dw_statistic = durbin_watson(model.resid)
    print(f"Durbin-Watson statistic: {dw_statistic}")
    
    # 3. Homoscedasticity
    plt.figure(figsize=(10, 6))
    sns.regplot(x=model.predict(), y=model.resid**2, lowess=True, line_kws={'color': 'red'})
    plt.xlabel("Predicted values")
    plt.ylabel("Squared residuals")
    plt.title("Homoscedasticity Check")
    plt.show()
    
    # 4. Normality of residuals
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    sns.histplot(model.resid, kde=True, ax=ax1)
    ax1.set_title("Histogram of Residuals")
    
    QQ = ProbPlot(model.resid)
    QQ.qqplot(line='45', ax=ax2)
    ax2.set_title("Q-Q plot of Residuals")
    plt.show()
    
    # Shapiro-Wilk test for normality
    _, p_value = stats.shapiro(model.resid)
    print(f"Shapiro-Wilk test p-value: {p_value}")
    
    # 5. No multicollinearity
    vif_data = pd.DataFrame()
    vif_data["Variable"] = X.columns
    vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
    print("\nVariance Inflation Factors:")
    print(vif_data)
    
    # 6. No influential outliers
    influence = model.get_influence()
    leverage = influence.hat_matrix_diag
    cook_distance = influence.cooks_distance[0]
    
    plt.figure(figsize=(10, 6))
    plt.scatter(leverage, cook_distance)
    plt.xlabel("Leverage")
    plt.ylabel("Cook's distance")
    plt.title("Influence Plot")
    plt.show()

# Assuming you have your OLS model results in 'results', X and y data
test_ols_assumptions(results, X, y)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from statsmodels.multivariate.manova import MANOVA
from statsmodels.stats.multicomp import pairwise_tukeyhsd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Assuming df_survey_cleaned is your dataframe
dependent_vars = ['PU_centered', 'PEOU_centered', 'HM_centered', 'PPV_centered', 
                  'SOI_centered', 'SEI_centered', 'AT_centered', 'PE_centered']
independent_var = 'AGE_num'

# 1. Post-hoc Analyses: Individual ANOVAs and Tukey's HSD
def post_hoc_analysis(df, dep_vars, indep_var):
    results = {}
    for var in dep_vars:
        # ANOVA
        groups = [group for name, group in df.groupby(indep_var)[var]]
        f_value, p_value = stats.f_oneway(*groups)
        
        # Tukey's HSD
        tukey = pairwise_tukeyhsd(df[var], df[indep_var])
        
        results[var] = {
            'ANOVA': {'F-value': f_value, 'p-value': p_value},
            'Tukey HSD': tukey
        }
    return results

post_hoc_results = post_hoc_analysis(df_survey_cleaned, dependent_vars, independent_var)

# Print results
for var, result in post_hoc_results.items():
    print(f"\n{var}:")
    print(f"ANOVA: F = {result['ANOVA']['F-value']:.4f}, p = {result['ANOVA']['p-value']:.4f}")
    print(result['Tukey HSD'])

# 2. Effect Size Calculation: Partial Eta Squared
def partial_eta_squared(df, dep_vars, indep_var):
    results = {}
    for var in dep_vars:
        groups = df.groupby(indep_var)[var]
        ss_between = sum(len(g) * (g.mean() - df[var].mean())**2 for _, g in groups)
        ss_total = sum((df[var] - df[var].mean())**2)
        eta_sq = ss_between / ss_total
        results[var] = eta_sq
    return results

effect_sizes = partial_eta_squared(df_survey_cleaned, dependent_vars, independent_var)
print("\nPartial Eta Squared (Effect Sizes):")
for var, eta_sq in effect_sizes.items():
    print(f"{var}: {eta_sq:.4f}")

# 3. Visualizations
# Box plots
plt.figure(figsize=(20, 15))
for i, var in enumerate(dependent_vars, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(x=independent_var, y=var, data=df_survey_cleaned)
    plt.title(f"{var} by Age Group")
plt.tight_layout()
plt.show()

# Canonical correlation plot (using PCA as a simplified approach)
X = df_survey_cleaned[dependent_vars]
y = df_survey_cleaned[independent_var]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(10, 8))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='viridis')
plt.colorbar(scatter)
plt.title('PCA of Dependent Variables Colored by Age Group')
plt.xlabel('First Principal Component')
plt.ylabel('Second Principal Component')
plt.show()

```

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
import matplotlib.pyplot as plt

# Assuming df_survey_cleaned is your DataFrame
# First, let's create a contingency table

def create_contingency_table(df, category_column, group_column):
    return pd.crosstab(df[group_column], df[category_column])

# Create a contingency table for AR usage categories and age cohorts
contingency_table = create_contingency_table(df_survey_cleaned, 'use_in_all_aspects', 'AGE_eng')

# Perform Chi-square test
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-square statistic: {chi2}")
print(f"p-value: {p_value}")

# Interpret the results
alpha = 0.05  # Significance level
if p_value < alpha:
    print("There is a significant difference in AR usage among different age cohorts.")
else:
    print("There is no significant difference in AR usage among different age cohorts.")

# Visualize the results
contingency_table.plot(kind='bar', stacked=True, figsize=(12, 6))
plt.title('AR Usage Categories by Age Cohort')
plt.xlabel('Age Cohort')
plt.ylabel('Count')
plt.legend(title='AR Usage Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# Post-hoc analysis: Standardized residuals
def standardized_residuals(observed, expected):
    return (observed - expected) / np.sqrt(expected)

std_residuals = standardized_residuals(contingency_table, expected)

# Plot heatmap of standardized residuals
plt.figure(figsize=(12, 8))
plt.imshow(std_residuals, cmap='RdYlBu', aspect='auto')
plt.colorbar(label='Standardized Residuals')
plt.title('Standardized Residuals: AR Usage Categories by Age Cohort')
plt.xlabel('AR Usage Category')
plt.ylabel('Age Cohort')
plt.xticks(range(len(contingency_table.columns)), contingency_table.columns, rotation=45, ha='right')
plt.yticks(range(len(contingency_table.index)), contingency_table.index)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
from itertools import combinations
from statsmodels.stats.multitest import multipletests

# Function to perform pairwise chi-square tests with Bonferroni correction
def pairwise_chi_square(df, group_column, category_column):
    # Get unique groups (age cohorts) and categories (AR usage)
    groups = df[group_column].unique()
    categories = df[category_column].unique()
    
    # Store p-values for each pairwise comparison
    p_values = []
    comparisons = []
    
    # Perform pairwise chi-square tests
    for group1, group2 in combinations(groups, 2):
        # Create a contingency table for each pair of groups
        group1_data = df[df[group_column] == group1][category_column].value_counts()
        group2_data = df[df[group_column] == group2][category_column].value_counts()
        
        # Align the indices (categories) of both groups
        combined_data = pd.DataFrame({'Group1': group1_data, 'Group2': group2_data}).fillna(0)
        
        # Perform Chi-square test on this subset
        chi2, p, dof, expected = chi2_contingency(combined_data)
        
        # Store results
        p_values.append(p)
        comparisons.append(f'{group1} vs {group2}')
    
    # Apply Bonferroni correction to p-values
    corrected_p_values = multipletests(p_values, method='bonferroni')[1]
    
    # Create a DataFrame with results
    results_df = pd.DataFrame({
        'Comparison': comparisons,
        'p-value': p_values,
        'Corrected p-value': corrected_p_values
    })
    
    return results_df

# Example usage with your DataFrame (replace with actual data)
pairwise_results = pairwise_chi_square(df_survey_cleaned, 'AGE_eng', 'use_in_all_aspects')

# Display results
print(pairwise_results)
```
